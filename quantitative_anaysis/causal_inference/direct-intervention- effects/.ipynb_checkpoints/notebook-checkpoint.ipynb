```python
import matplotlib.pyplot as plt
import numpy as np
import networkx as nx
import matplotlib.cm as cm
import matplotlib.colors as mcolors
```

# Generate our city

```python
city = SynthCity(n_people=5000, seed=0)
baseline_df = city.sample()
baseline_df.head()
```

## Ground-truth intervention: what happens if min wage goes to $15?


```python
treated_city = city.do(min_wage=15.0)
treated_df = treated_city.sample()
```

```python
fig, ax = plt.subplots(figsize=(7, 4))
ax.hist(baseline_df["income"], bins=50, alpha=0.6, label="baseline ($7.25 min wage)")
ax.hist(treated_df["income"], bins=50, alpha=0.6, label="treated ($15 min wage)")
ax.set_xlabel("annual income")
ax.set_ylabel("count")
ax.legend()
ax.set_title("Income distribution: observational vs. intervention")
fig.savefig("../results/figures/income_distribution_shift.png", dpi=150, bbox_inches="tight")
```

## Dose-response curve: sweep min_wage and plot the true ATE at each level


```python
wage_levels = np.arange(7.25, 25, 1.0)
mean_incomes = [city.do(min_wage=w).sample()["income"].mean() for w in wage_levels]
```

```python
fig, ax = plt.subplots(figsize=(7, 4))
ax.plot(wage_levels, mean_incomes, marker="o")
ax.set_xlabel("min_wage")
ax.set_ylabel("mean income")
ax.set_title("True dose-response curve: min_wage → mean income")
fig.savefig("../results/figures/min_wage_dose_response.png", dpi=150, bbox_inches="tight")
```

## The DAG

```python
fig, ax = plt.subplots(figsize=(8, 6))
pos = nx.spring_layout(city.dag, seed=0)
nx.draw(city.dag, pos, ax=ax, with_labels=True, node_color="lightblue",
        node_size=2000, font_size=8, arrowsize=15)
fig.savefig("../results/figures/dag_structure.png", dpi=150, bbox_inches="tight")
```

```python
def compute_shifts(baseline_df, treated_df):
    shifts = {}
    for col in baseline_df.columns:
        base_mean, treat_mean = baseline_df[col].mean(), treated_df[col].mean()
        base_std = baseline_df[col].std()
        delta = treat_mean - base_mean
        effect_size = delta / base_std if base_std > 0 else 0.0
        shifts[col] = {"delta": delta, "effect_size": effect_size}
    return shifts
```

```python
shifts = compute_shifts(baseline_df, treated_df)

# reuse ONE layout across plots so nodes don't jump around between figures
pos = nx.spring_layout(city.dag, seed=0)

intervened_nodes = {"min_wage"}  # whatever you passed to .do(...)
effect_sizes = {n: shifts[n]["effect_size"] for n in city.dag.nodes}
max_abs = max(abs(v) for k, v in effect_sizes.items() if k not in intervened_nodes) or 1

norm = mcolors.TwoSlopeNorm(vmin=-max_abs, vcenter=0, vmax=max_abs)
cmap = cm.RdBu_r
node_colors = [
    "gold" if n in intervened_nodes else cmap(norm(effect_sizes[n]))
    for n in city.dag.nodes
]
```

```python
fig, ax = plt.subplots(figsize=(9, 7))
nx.draw_networkx_edges(city.dag, pos, ax=ax, arrowsize=15, alpha=0.5)
nx.draw_networkx_nodes(city.dag, pos, ax=ax, node_color=node_colors,
                        node_size=2200, edgecolors="black", linewidths=1)
nx.draw_networkx_labels(city.dag, pos, ax=ax, font_size=8)

for n, (x, y) in pos.items():
    ax.text(x, y - 0.08, f"Δ={shifts[n]['delta']:+.2f}", fontsize=7,
             ha="center", color="dimgray")

sm = cm.ScalarMappable(norm=norm, cmap=cmap)
sm.set_array([])
fig.colorbar(sm, ax=ax, shrink=0.7, label="standardized effect size")
ax.set_title("DAG shift under do(min_wage=15.0)  [gold = intervened node]")
ax.axis("off")
fig.savefig("../results/figures/dag_shift_min_wage.png", dpi=150, bbox_inches="tight")
```

* The gold node is the intervention; minimum wage = 15.
  
* The standardized effect size (delta / baseline_std) at every other node, from blue to red intensity,compares the  
shift magnitude across variables(ie. income in the tens of thousands vs. health as a standardized score).

* The White/neutral nodes (tax_rate, transit_subsidy) should show exactly zero are controls that show the intervention didn't leak anywhere it shouldn't.

R: The effect of the intervention fades as it propagates: income gets hit hardest, mobility_access and health show smaller effect sizes, and education/ability/parental_ses should show essentially zero since nothing flows backward up the DAG 

## True ATE effect

```python
true_ate = treated_df["income"].mean() - baseline_df["income"].mean()

fig, ax = plt.subplots(figsize=(7, 4))
ax.hist(baseline_df["income"], bins=50, alpha=0.6, label="baseline ($7.25 min wage)")
ax.hist(treated_df["income"], bins=50, alpha=0.6, label="treated ($15 min wage)")
ax.axvline(baseline_df["income"].mean(), color="C0", linestyle="--")
ax.axvline(treated_df["income"].mean(), color="C1", linestyle="--")
ax.set_xlabel("annual income")
ax.set_ylabel("count")
ax.legend()
ax.set_title(f"True ATE = ${true_ate:,.0f}")
fig.savefig("../results/figures/income_distribution_shift.png", dpi=150, bbox_inches="tight")

print(f"true_ate: {true_ate:,.2f}")
```

### True ATE effect, averag

```python
n_seeds = 20
ates = []
for s in range(n_seeds):
    c = SynthCity(n_people=5000, seed=s)
    b = c.sample()
    t = c.do(min_wage=15.0).sample()
    ates.append(t["income"].mean() - b["income"].mean())

true_ate_stable = np.mean(ates)
true_ate_std = np.std(ates)
print(f"true_ate ≈ {true_ate_stable:,.2f} ± {true_ate_std:,.2f}")
```
